```python
# Array Functions , max,min , mean, avg
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
                      
```

```python
# max find each rows and min 
#axis = 0 for col, and  axis = 1 for rows

print("rows wise max : ",np.max(a1, axis =1))
print("cols wise max : ",np.max(a1, axis =0))
np.min(a1, axis = 1)
print("cols wise sum : ",np.sum(a1, axis =0))
print("rows wise product : ",np.prod(a1, axis =1))
```

```python
# find mean/median/std/var

print("cols wise mean: ",np.mean(a1,axis = 0))
print("rows wise mean: ",np.mean(a1,axis = 1))
print("rows wise variance: ",np.var(a1,axis = 1))
print("rows wise standard deviation: ",np.std(a1,axis = 1))

```

```python
# round/floor/ceil
a = np.random.random((2,3))*100
print(a,"\n")
print("round: " ,np.round(a))    #1.59 = 2 hoga and 1.49 = 1 hoga (.5 =< se bada h to round me 1 consider karega  
print("floor: ",np.floor(a))     #cosider the left value only mean before decimal
print("ceil: ",np.ceil(a))        #1. 59 = 2 , 1.03 = 2 mean decimal hai to one add kar do
```

```python
arr[1:,1:,]    # 1st(index)/2nd rows se end row tak ke 1st(index)/2nd colums se sab
```
